```python
from math import sqrt
n = 14
print(f'Square Root of {n} is {sqrt(n)}')
print(f'Cube of {n} is {pow(n,3)}')
def is_prime(n):
	for _ in range(2,int(sqrt(n))+1):
		if(n%_==0):
			return(f'{n} is not prime')
		else:
			return(f'{n} is prime')
print(f'{is_prime(n)}')
def fact(n, memo={}):
	if n in memo:
		return memo[n]
	elif n == 0:
		return 1
	elif n == 1 or n == 2:
		return n
	else:
		result = n * fact(n-1, memo)
	memo[n] = result
	return result
print(f'Factorial of {n} is {fact(n)}')
def prime_fact(n):
	facts = []
	while n%2==0:
		facts.append(2)
		n //= 2

	for i in range(3, int(sqrt(n))+1, 2):
		while n%i==0:
			facts.append(i)
			n //= i

	if n > 2:
		facts.append(n)

	return facts
print(f'Prime Factors of {n} is {prime_fact(n)}')
```

```python
from math import sqrt
def ball_collide(b1, b2):
	x1, y1, r1 = b1
	x2, y2, r2 = b2
	dist = sqrt(pow(x2-x1, 2) + pow(y2-y1, 2))
	r = r1 + r2
	if dist <= r:
		return True
	return False

ball1 = (2,3,4)
ball2 = (5,5,2)
ball_collide(ball1, ball2)
```

```python
def fact(n, memo=None):
    if memo is None:
        memo = {}

    if not isinstance(n, int):
        raise ValueError(f'Cannot compute factorial, positive integer required instead got {type(n)}')
    elif n < 0:
        raise ValueError('Cannot compute factorial of negative number')
    elif n == 1 or n == 0:
        return 1
    elif n == 2:
        return n
    elif n in memo:
        return memo[n]
    else:
        res = n * fact(n-1, memo)
        memo[n] = res
        return res

fact('-5')
```

```python
class Complex:
	def __init__(self, real, imag):
		self.real = real
		self.imag = imag
	
	def __add__(self, other):
		real = self.real + other.real
		imag = self.imag + other.imag
		return Complex(real, imag)

	def __str__(self):
		return f'({self.real} {'+' if self.imag > 0 else '-'} {abs(self.imag)}i)'

def add_complex(cnums):
	t = cnums[0]
	for i in range(1, len(cnums)):
		t += cnums[i]
	return t

nums = []
n = int(input('Enter no. of Complex Numbers : '))
for _ in range(n):
	real = int(input('Enter Real Part : '))
	imag = int(input('Enter Imag Part : '))
	nums.append(Complex(real, imag))

print(*nums, sep=' + ', end=' = ')
print(add_complex(nums))
```
